## Section 1: Aggregate Functions

#### 1 Calculate the total number of students in each department. 

```python
%%sql
Select d.name as deoartment_name , count( distinct sc.student_ID) as total_student
from courses c,department d, student_courses sc
where c.department_ID=d.Dep_ID and sc.courses_ID=c.cour_ID
group by d.Dep_ID ,d.name
```


#### 2 Find the average GPA of students in each department. 

```python
%%sql
select d.name as department_name, cast( avg(s.GPA) as decimal(10,2)) as AVG_GPA 
from student s,courses c,department d, student_courses sc
where c.department_ID=d.Dep_ID and sc.student_ID=s.ST_ID and c.cour_ID=sc.courses_ID 
group by d.name,d.name
```

#### 3 Determine the maximum and minimum GPA of students in each department

```python
%%sql
select d.name as Department_name,max(s.GPA) as Max_GPA,min(GPA) aS Min_GPA 
from student s,department d, courses c,student_courses sc
where c.department_ID=d.Dep_ID and s.ST_ID=sc.student_ID and c.cour_ID=sc.courses_ID 
group by d.Dep_ID,d.name
```

#### 4 Calculate the average GPA for students in each enrollment year

```python
%%sql
select s.enrollmentyear , cast( AVG(s.GPA) as decimal(10,2)) as AVG_GPA 
from student s
group by s.enrollmentyear
```

#### 5 Find the total number of students participating in research projects for each projec

```python
%%sql
select r.title,count(s.ST_ID)as total_student
from student s,Research_Project r, project_involvement p
where s.ST_ID=p.Student_ID and r.Research_Project_ID= p.Research_Project_ID
group by r.title ,r.Research_Project_ID
```

### Section 2: Joins

#### 1 Retrieve the names of all students along with the names of the courses they are enrolled in.

```python
%%sql
select s.name as student_name,c.name as courses_name from student s,courses c , student_courses sc
where s.ST_ID=sc.student_ID and c.cour_ID=sc.courses_ID
```

#### 2 Find the names of the professors who teach courses offered by the Computer Science department.

```python
%%sql
select p.name as professor_name,d.name as department_name from professor p , department d , professor_department pd
where p.professor_ID=pd.professor_ID and d.Dep_ID=pd.department_ID
and d.name= 'Computer Science'
```

#### 3 Retrieve the names of all students and the names of the courses they are enrolled in, sorted by course name.

```python
%%sql
select s.name as student_Name,c.name as courses_Name from student s,courses c, student_courses sc 
where s.ST_ID=sc.student_ID and c.cour_ID=sc.courses_ID order by c.name asc
```

#### 4 For each department, display:  Department name  Professor name  Academic rank Include departments that do not currently have professors

```python
%%sql
select distinct d.name as department_name ,p.name professor_name from department d left join professor_department pd on d.Dep_ID=pd.department_ID
left join professor p on p.professor_ID=pd.professor_ID
```

#### 5 For each research project, display:  Project ID  Project name  Department name  Student name Include projects that do not have any students assignedto them.

```python
%%sql
select rp.Research_Project_ID,rp.title as Researsh_title,d.name as department_name ,s.name as student_name  from Research_Project rp 
left join project_involvement pi on rp.Research_Project_ID=pi.Research_Project_ID
left join student s on s.ST_ID =pi.Student_ID
left join professor_department pd on pi.professor_ID=pd.professor_ID
left join department d on d.Dep_ID=pd.department_ID
```

## Section 3: UNION Operations 


#### 1 Combine the names of all students and professors into a single list.

```python
%%sql
select s.name from student s
union
select p.name from professor p
```

#### 2 Create a single list containing:  Student names whose GPA is greater than or equal to 3.0  Professor names 

```python
%%sql
select s.name from student s where s.GPA >=3.0
union
select p.name from professor p
```

#### 3 Combine the names of all:  Students  Professors  Administrative Staff

```python
%%sql
select s.name from student s
union
select p.name from professor p
union
select f.name from staff f 
```

#### --Find the department with the highest number of students. 

```python
%%sql
 select top 1  d.name as department_name ,count(s.ST_ID)as count_student  from student s, department d, courses c, student_courses sc
 where s.ST_ID=sc.student_ID and c.cour_ID=sc.courses_ID and d.Dep_ID=c.department_ID group by d.name order by count_student desc
```

#### Find the course with the highest number of enrolled students. 

```python
%%sql
select top 1 c.name as courses_name ,count(s.ST_ID) as count_student  from student s ,courses c ,student_courses sc
where s.ST_ID= sc.student_ID and c.cour_ID=sc.courses_ID group by c.name order by count_student desc
```

#  Views 

#### 1. Create a view that shows each department’s ID, department name, and the 
#### full name of its professor/manager. 

```python
%%sql
create or alter view dep_12 
as
select d.Dep_ID,d.name as department_name,p.name as professor_name from department d, professor p , professor_department pd
where d.Dep_ID=pd.department_ID and p.professor_ID=pd.professor_ID
```

```python
%%sql
select * from dep_12
```

#### 2. Create a view that displays each department name along with the names of 
#### the courses offered by that department.

```python
%%sql
 create or alter view dep_cour
 as
select d.name as department_name ,c.name as coueses_name from department d, courses c 
where d.Dep_ID=c.department_ID 
```

```python
%%sql
select * from dep_cour
```

####  Create a view that displays the full details of all students along with the courses they are enrolled in, including the enrollment date and grade.

```python
%%sql
create or alter view st_cour
as
 select s.* ,c.name as courses_name ,sc.enrollment_Data ,sc.grate   from student s ,courses c , student_courses sc
 where s.ST_ID=sc.student_ID and c.cour_ID=sc.courses_ID
```

```python
%%sql
select * from st_cour
```

#### Create a view to retrieve the names of all students enrolled in department/major "Computer Science" who have a GPA greater than or equal to 3.0.

```python
%%sql
 create or alter view smg
as
select s.name as student_name  from student s  where major= 'CS' and GPA >= 3.0
```

```python
%%sql
select * from smg
```
